## Cell 1 — Import Required Libraries


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

pd.set_option('display.max_columns', None)
```

## Cell 2 — Define Dataset Path


```python
file_path = r"C:\Users\bts\OneDrive\Desktop\SaiKet Systems\dataset\Telco_Customer_Churn_Dataset.csv"
print("Dataset path:", file_path)
```

## Cell 3 — Load the Dataset


```python
df = pd.read_csv(file_path)
print("Dataset loaded successfully!")
print("Shape:", df.shape)
df.head()
```

## Cell 4 — Initial Dataset Exploration


```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("\nColumn names:")
print(df.columns.tolist())
```

## Cell 5 — View Dataset Information


```python
df.info()
```

## Cell 6 — Statistical Summary


```python
df.describe(include="all").T
```

## Cell 7 — Check Missing Values


```python
missing_values = df.isnull().sum().sort_values(ascending=False)
missing_percentage = (df.isnull().mean() * 100).sort_values(ascending=False)

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
})

missing_summary[missing_summary["Missing Values"] > 0]
```

## Cell 8 — Check Duplicate Records


```python
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)
```

## Cell 9 — Inspect Unique Values of Categorical Columns


```python
categorical_columns = df.select_dtypes(
    include=["object", "str", "category"]
).columns

print("Categorical columns:")
print(categorical_columns.tolist())

for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].unique())
```

## Cell 10 — Clean TotalCharges and Handle Invalid Values


```python
if "TotalCharges" in df.columns:
    df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

print("Missing values after numeric conversion:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

## Cell 11 — Handle Missing Values


```python
# Numerical missing values → median
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())

# Categorical missing values → mode
categorical_columns = df.select_dtypes(
    include=["object", "str", "category"]
).columns

for col in categorical_columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

print("Remaining missing values:", df.isnull().sum().sum())
```

## Cell 12 — Remove Duplicate Rows


```python
before = len(df)
df = df.drop_duplicates().reset_index(drop=True)
after = len(df)

print("Rows before removing duplicates:", before)
print("Rows after removing duplicates:", after)
print("Duplicates removed:", before - after)
```

## Cell 13 — Separate Target Variable from Features


```python
target_column = "Churn"

if target_column not in df.columns:
    raise KeyError(
        f"Target column '{target_column}' was not found. "
        f"Available columns: {df.columns.tolist()}"
    )

X = df.drop(columns=[target_column]).copy()
y = df[target_column].copy()

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nTarget distribution:")
print(y.value_counts())
```

## Cell 14 — Remove Customer ID from Features


```python
# Customer ID is an identifier and is not useful as a predictive feature.
id_columns = [
    col for col in ["customerID", "customerId", "CustomerID"]
    if col in X.columns
]

if id_columns:
    X = X.drop(columns=id_columns)
    print("Removed identifier column(s):", id_columns)
else:
    print("No customer ID column found.")
```

## Cell 15 — Encode Binary Target Variable


```python
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y)

print("Target classes:", label_encoder.classes_)
print("Encoded target values:", np.unique(y))
```

## Cell 16 — Encode Categorical Feature Variables


```python
categorical_features = X.select_dtypes(
    include=["object", "str", "category"]
).columns.tolist()

print("Categorical feature columns:")
print(categorical_features)

# One-hot encoding
X = pd.get_dummies(
    X,
    columns=categorical_features,
    drop_first=True,
    dtype=int
)

print("Shape after encoding:", X.shape)
X.head()
```

## Cell 17 — Verify Preprocessed Data


```python
print("Missing values in X:", X.isnull().sum().sum())
print("Missing values in y:", pd.isnull(y).sum())
print(
    "All feature columns numeric:",
    all(np.issubdtype(dtype, np.number) for dtype in X.dtypes)
)
print("\nFinal feature shape:", X.shape)
X.head()
```

## Cell 18 — Split Data into Training and Testing Sets


```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

## Cell 19 — Verify Target Distribution After Split


```python
print("Training target distribution:")
print(pd.Series(y_train).value_counts(normalize=True).sort_index())

print("\nTesting target distribution:")
print(pd.Series(y_test).value_counts(normalize=True).sort_index())
```
